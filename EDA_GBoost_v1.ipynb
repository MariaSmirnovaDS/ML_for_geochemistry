**Разбиение на train, val, test**

```python
train_wells = ['A1', 'A3', 'A4', 'B1', 'C1', 'C4', 'C5', 'C6', 'D1']
val_wells   = ['A2', 'E1']
test_wells  = ['C2', 'C3', 'C7']

train = df_filtered[df_filtered['Well'].isin(train_wells)].copy()
val   = df_filtered[df_filtered['Well'].isin(val_wells)].copy()
test  = df_filtered[df_filtered['Well'].isin(test_wells)].copy()

print(f"Train: {len(train)} проб, {train['Well'].nunique()} скважин, {train['Strat'].nunique()} пачек")
print(f"Val:   {len(val)} проб, {val['Well'].nunique()} скважин, {val['Strat'].nunique()} пачек")
print(f"Test:  {len(test)} проб, {test['Well'].nunique()} скважин, {test['Strat'].nunique()} пачек")
```

**CATBOOST. Вариант 1**

```python
from catboost import CatBoostClassifier
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

X_train, y_train = train[elements], train['Strat']
X_val, y_val     = val[elements], val['Strat']
X_test, y_test   = test[elements], test['Strat']

# Модель с подбором параметров через сетку
from sklearn.model_selection import ParameterGrid

param_grid = {
    'depth': [4, 6, 8],
    'learning_rate': [0.03, 0.1, 0.3],
    'iterations': [500, 1000],
}

best_score = 0
best_params = {}
best_model = None

for params in ParameterGrid(param_grid):
    model = CatBoostClassifier(
        **params,
        loss_function='MultiClass',
        eval_metric='TotalF1',
        random_seed=42,
        verbose=0,         
        early_stopping_rounds=50
    )
    model.fit(X_train, y_train, eval_set=(X_val, y_val), use_best_model=True)
    
    # F1 macro на val
    from sklearn.metrics import f1_score
    y_val_pred = model.predict(X_val)
    score = f1_score(y_val, y_val_pred, average='macro')
    
    if score > best_score:
        best_score = score
        best_params = params
        best_model = model
        print(f"Новый лучший: F1_macro={score:.4f}, params={params}")

print(f"\nЛучшие параметры: {best_params}")
print(f"Лучший F1 macro на val: {best_score:.4f}")

# Финальная оценка на test
y_pred = best_model.predict(X_test)

# F1 по каждому классу + суммарные (macro и weighted)
print("\n=== Classification Report (Test) ===")
print(classification_report(y_test, y_pred, digits=4))

# Матрица ошибок
plt.figure(figsize=(10, 8))
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=range(12), yticklabels=range(12))
plt.xlabel('Предсказание')
plt.ylabel('Истина')
plt.title('Матрица ошибок (Test)')
plt.tight_layout()
plt.show()

# Важность признаков
importance = best_model.get_feature_importance()
feat_imp = sorted(zip(elements, importance), key=lambda x: x[1], reverse=True)

plt.figure(figsize=(8, 6))
sns.barplot(x=[f[1] for f in feat_imp], y=[f[0] for f in feat_imp], palette='viridis')
plt.xlabel('Важность')
plt.title('Важность признаков')
plt.tight_layout()
plt.show()
```

**Важность признаков по каждому классу**

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from catboost import Pool

# SHAP-значения для test
# специальный формат CatBoost, объединяет признаки и целевую переменную в один объект.
pool_test = Pool(X_test, y_test)
shap_values = best_model.get_feature_importance(
    pool_test, 
    type='ShapValues'    # SHAP считает насколько каждый признак сдвинул предсказание для конкретной пробы от базы
)

# shap_values: (n_samples, n_classes, n_features + 1)
# последний столбец — базовое значение, его отбрасываем
shap_per_class = shap_values[:, :, :-1]  # (n_samples, n_classes, n_features)

# Среднее |SHAP| по каждому классу и признаку
mean_shap = np.abs(shap_per_class).mean(axis=0)  # (n_classes, n_features)

# Тепловая карта: классы × признаки
plt.figure(figsize=(12, 8))
sns.heatmap(
    mean_shap,
    annot=True, fmt='.3f',
    cmap='YlOrRd',
    xticklabels=elements,
    yticklabels=[f'Пачка {i}' for i in range(12)],
    cbar_kws={'label': 'Среднее |SHAP|'}
)
plt.xlabel('Признаки (элементы)')
plt.ylabel('Классы (пачки)')
plt.title('Важность признаков по каждому классу (SHAP)')
plt.tight_layout()
plt.show()
```

```python
for i in range(mean_shap.shape[0]):
    significant = [(elements[j], mean_shap[i, j])
                   for j in range(len(elements))
                   if mean_shap[i, j] > 0.4]
    significant.sort(key=lambda x: -x[1])
    print(f"Пачка {i}:")
    for elem, v in significant:
        print(f"  {elem}: {v:.3f}")
    print()
```

**Вывод:** F1 macro = 0.61, weighted = 0.66 — модель улавливает закономерности, но половина пачек распознаётся плохо.

**Выгрузка предсказаний**

```python
all_data = pd.concat([train, val, test], ignore_index=True)
X_all = all_data[elements]

# Предсказываем классы
all_data['Pred_Strat'] = best_model.predict(X_all).ravel()

# Вероятности для каждого класса
proba = best_model.predict_proba(X_all)
proba_cols = [f'Prob_{i}' for i in range(proba.shape[1])]
all_data[proba_cols] = proba

# Метка выборки
all_data['Split'] = ''
all_data.loc[:len(train)-1, 'Split'] = 'train'
all_data.loc[len(train):len(train)+len(val)-1, 'Split'] = 'val'
all_data.loc[len(train)+len(val):, 'Split'] = 'test'

# Флаг ошибки
all_data['Correct'] = (all_data['Strat'] == all_data['Pred_Strat']).astype(int)

# Выгрузка
all_data.to_csv('Result/classification_results_Catboost_v1.csv', index=False, encoding='utf-8')
print(f'Выгружено {len(all_data)} проб в classification_results.csv')
print(f'  train: {len(train)}, val: {len(val)}, test: {len(test)}')
print(f'  Колонки: {list(all_data.columns)}')
```

**Вывод:** Это высокая, но не экстремальная точность — модель не просто зазубрила трейн, но и выучила реальные закономерности. 0.97 говорит о том, что 3% проб модель осознанно не может разделить — скорее всего, это переходные границы между пачками, где состав элементов действительно похож. Разрыв train→test в 0.30 — это переобучение, но для геологических данных с 12 классами и разбиением по скважинам это ожидаемо. Пробы из одной скважины скоррелированы между собой, а модель тестируется на совершенно новых скважинах — это объективно тяжёлая задача.

Можно попробовать немного сгладить разрыв:

- Увеличить l2_leaf_reg (3 или 5 вместо 1) — сильнее ограничит подгонку
- Уменьшить depth (4 вместо 6) — проще деревья, меньше шансов запомнить
- Уменьшить learning_rate и увеличить iterations — учиться медленнее и тщательнее
