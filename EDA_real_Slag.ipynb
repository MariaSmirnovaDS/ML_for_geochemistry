```python
import pandas as pd
import os
import glob
```

```python
folder = 'Data/SLAG/'
sheet_name = 'X-MET'

needed_cols = ['Скважина', '№ пробы', 'Глубина_м', 'Стратиграф.']
elements = ['Mg', 'Al', 'Si', 'P', 'S', 'K', 'Ca', 'Ti', 'Mn', 'Fe',
            'Ni', 'Cu', 'Zn', 'Se', 'Rb', 'Sr', 'Zr', 'Mo']

all_data = []
files = glob.glob(os.path.join(folder, '*.xls*'))

print(f"найдено файлов: {len(files)}\n")

for file_path in files:
    filename = os.path.basename(file_path)
    well_name = filename.split('-')[0].split('_')[0]
    
    try:
        df = pd.read_excel(file_path, sheet_name=sheet_name, header=0)
        
        needed = needed_cols + elements
        missing = [c for c in needed if c not in df.columns]
        if missing:
            print(f"  {well_name}: пропущены {missing}")
            continue
        
        df = df[needed].copy()
        df['Well'] = well_name
        all_data.append(df)
        print(f"  {well_name}: {len(df)} проб")
        
    except Exception as e:
        print(f"  {well_name}: ошибка — {e}")

df_slag = pd.concat(all_data, ignore_index=True)

df_slag = df_slag.rename(columns={
    'Скважина': 'Well_file',
    '№ пробы': 'Sample',
    'Глубина_м': 'Depth',
    'Стратиграф.': 'Stratigraphy'
})

meta_cols = ['Well', 'Well_file', 'Sample', 'Depth', 'Stratigraphy']
df_slag = df_slag[meta_cols + elements]

print(f"\nвсего проб: {len(df_slag)}")
print(f"скважин: {df_slag['Well'].nunique()}")
print(f"колонки: {df_slag.columns.tolist()}")
```

**Подсчет нулей по всем признакам**

```python
# Подсчет нулей по всем признакам
zero_counts = df_slag[elements].isnull().sum() + (df_slag[elements] == 0).sum()

# Выводим результат
print("Количество нулевых значений по каждому элементу:")
print(zero_counts)

# Общее количество нулей
total_zeros = zero_counts.sum()
print(f"\nОбщее количество нулевых значений: {total_zeros}")

# Процент нулевых значений
total_values = len(df_slag) * len(elements)
zero_percentage = (total_zeros / total_values) * 100
print(f"Процент нулевых значений: {zero_percentage:.2f}%")
```

**Подсчет значений ниже LOD**

```python
# LOD для каждого элемента
lod = {
    'Mg': 0.1848625,
    'Al': 0.0927875,
    'Si': 0.2984125,
    'Ca': 0.077025,
    'S': 0.3082,
    'K': 0.0271375,
    'Ti': 0.003,
    'Mn': 0.0024375,
    'Fe': 0.09165,
    'Ni': 0.0032625,
    'Cu': 0.00335,
    'Zn': 0.0022375,
    'Rb': 0.0003125,
    'Sr': 0.0013125,
    'Zr': 0.00065,
    'Mo': 0.00105,
    'P': 0.016325,
    'Se': 0.0002375
}
```

```python
below_lod = pd.Series({e: ((df_slag[e] > 0) & (df_slag[e] < lod[e])).sum() for e in elements})

print("Количество значений ниже LOD (без нулей):")
print(below_lod)

print(f"\nВсего ниже LOD: {below_lod.sum()}")
print(f"Процент: {(below_lod.sum() / (len(df_slag) * len(elements))) * 100:.2f}%")
```

**Поиск минимальных значений по элементам**

```python
min_values = pd.Series({e: df_slag.loc[df_slag[e] > 0, e].min() for e in elements})

print("Минимальные значения (исключая 0):")
print(min_values)
```

**Минимальное или LOD, если он меньше**

```python
result = pd.Series({e: min(min_values[e], lod[e]) for e in elements})

print("Минимум (меньше из минимального значения и LOD):")
print(result)
```

**Замена нулей на 2/3 LOD (result)**

```python
# Замена нулей на 2/3 LOD
for el in elements:
    replacement = (2/3) * result[el]
    df_slag[el] = df_slag[el].replace(0, replacement)

# Проверка
print("Нулей после замены:")
print((df_slag[elements] == 0).sum())
```

**Создание новых признаков по соотношениям элементов**

```python
df_slag['(Ca+Mg)/(Al+Si)'] = (df_slag['Ca'] + df_slag['Mg']) / (df_slag['Al'] + df_slag['Si'])
df_slag['Sr/Ca'] = df_slag['Sr'] / df_slag['Ca']
df_slag['Mn/Ca'] = df_slag['Mn'] / df_slag['Ca']

# терригенное вещество
df_slag['(Al+K)/(Ca+Mg+Si)'] = (df_slag['Al'] + df_slag['K']) / (df_slag['Ca'] + df_slag['Mg'] + df_slag['Si'])
df_slag['Zr/Al'] = df_slag['Zr'] / df_slag['Al']
df_slag['Zr/Si'] = df_slag['Zr'] / df_slag['Si']
df_slag['K/Al'] = df_slag['K'] / df_slag['Al']
df_slag['Ti/(Zr+Ti)'] = df_slag['Ti'] / (df_slag['Zr'] + df_slag['Ti'])

# морская биопродуктивность и кислородный режим
df_slag['Si/Al'] = df_slag['Si'] / df_slag['Al']
df_slag['P/Al'] = df_slag['P'] / df_slag['Al']
df_slag['Ni/Fe'] = df_slag['Ni'] / df_slag['Fe']
df_slag['Ni/(Si+Ca+Mg)'] = df_slag['Ni'] / (df_slag['Si'] + df_slag['Ca'] + df_slag['Mg'])
df_slag['S/(Rb+Zr)'] = df_slag['S'] / (df_slag['Rb'] + df_slag['Zr'])
df_slag['Mo/Al'] = df_slag['Mo'] / df_slag['Al']
df_slag['Mo/(Mn+Mo)'] = df_slag['Mo'] / (df_slag['Mn'] + df_slag['Mo'])
df_slag['Mn/Al'] = df_slag['Mn'] / df_slag['Al']

# сульфидность
df_slag['Fe/S'] = df_slag['Fe'] / df_slag['S']
df_slag['Ni/S'] = df_slag['Ni'] / df_slag['S']

# железистость
df_slag['Fe/(Ca+Mg+Al)'] = df_slag['Fe'] / (df_slag['Ca'] + df_slag['Mg'] + df_slag['Al'])
df_slag['Ti/(Si+Ca+Mg)'] = df_slag['Ti'] / (df_slag['Si'] + df_slag['Ca'] + df_slag['Mg'])
df_slag['Fe/(Mn+Fe)'] = df_slag['Fe'] / (df_slag['Mn'] + df_slag['Fe'])

# прочие
df_slag['(Al+K+Ti)/Si'] = (df_slag['Al'] + df_slag['K'] + df_slag['Ti']) / df_slag['Si']
df_slag['(Al+K+Ti)/S'] = (df_slag['Al'] + df_slag['K'] + df_slag['Ti']) / df_slag['S']
df_slag['(Ti+Rb+Zr)/S'] = (df_slag['Ti'] + df_slag['Rb'] + df_slag['Zr']) / df_slag['S']
df_slag['S/Ti'] = df_slag['S'] / df_slag['Ti']
df_slag['Ni/Al'] = df_slag['Ni'] / df_slag['Al']
df_slag['Zn/Al'] = df_slag['Zn'] / df_slag['Al']
df_slag['Se/Si'] = df_slag['Se'] / df_slag['Si']
df_slag['(Ca+Mg)/Al'] = (df_slag['Ca'] + df_slag['Mg']) / df_slag['Al']
df_slag['Se/Ti'] = df_slag['Se'] / df_slag['Ti']
df_slag['(10*Rb+Ti+10*Zr)/Se*1000'] = (10 * df_slag['Rb'] + df_slag['Ti'] + 10 * df_slag['Zr']) / df_slag['Se'] * 1000
```

**Создание относительных признаков**

```python
features = [
    'Mg', 'Al', 'Si', 'P', 'S', 'K', 'Ca', 'Ti', 'Mn', 'Fe',
    'Ni', 'Cu', 'Zn', 'Se', 'Rb', 'Sr', 'Zr', 'Mo',
    '(Ca+Mg)/(Al+Si)',
    'Sr/Ca',
    'Mn/Ca',
    '(Al+K)/(Ca+Mg+Si)',
    'Zr/Al',
    'Zr/Si',
    'K/Al',
    'Ti/(Zr+Ti)',
    'Si/Al',
    'P/Al',
    'Ni/Fe',
    'Ni/(Si+Ca+Mg)',
    'S/(Rb+Zr)',
    'Mo/Al',
    'Mo/(Mn+Mo)',
    'Mn/Al',
    'Fe/S',
    'Ni/S',
    'Fe/(Ca+Mg+Al)',
    'Ti/(Si+Ca+Mg)',
    'Fe/(Mn+Fe)',
    '(Al+K+Ti)/Si',
    '(Al+K+Ti)/S',
    '(Ti+Rb+Zr)/S',
    'S/Ti',
    'Ni/Al',
    'Zn/Al',
    'Se/Si',
    '(Ca+Mg)/Al',
    'Se/Ti',
    '(10*Rb+Ti+10*Zr)/Se*1000'
]
```

**Расчет скользящих градиентов Δxi = xi - xi-1 по каждой скважине (сортировка по Depth)**

```python
# Сортировка df по скважине и глубине
df_sorted = df_slag.sort_values(['Well', 'Depth']).reset_index(drop=True)

# Для каждого признака разность с предыдущей пробой в той же скважине
gradient_features = []
for col in features:
    grad_col = f'dΔ_{col}'
    df_sorted[grad_col] = df_sorted.groupby('Well')[col].diff()
    gradient_features.append(grad_col)

# Исходный порядок
df = df_sorted

print(f"Добавлено градиентных признаков: {len(gradient_features)}")
```

**Создание относительных градиентов rΔxi = xi / (xi-1 + epsilon)**

```python
df = df.sort_values(['Well', 'Depth']).reset_index(drop=True)

rgrad_features = []
for col in features:
    rgrad_col = f'rΔ_{col}'
    prev = df.groupby('Well')[col].shift(1)
    epsilon = df[col].median() * 0.01
    df[rgrad_col] = df[col] / (prev + epsilon)
    rgrad_features.append(rgrad_col)

print(f"Добавлено в df: {len(rgrad_features)} признаков")
print(f"Всего колонок: {len(df.columns)}")
```
